```python
print("=== Sistema da nave ===")
print("Digite os valores atuais para verificar se estao dentro do permitido.\n")

dados_minimos = {
    "energia": 20,
    "temperatura_externa": (0, 45),
    "temperatura_interna": (15, 35),
    "pressao_tanque": (100, 145),  # dado retirado do google sobre a pressao minima e maxima.
    "integridade_estrutural": 1,   # 0 = NÃO / 1 = SIM
    "status_modulos_criticos": 1,  # 0 = FALHA / 1 = OK
}

especificacoes_nave = {
    "capacidade_total_kwh": 100,
    "consumo_decolagem_kwh": 20,
    "consumo_sistemas_kw": 2,
    "perda_energetica_percent": 5,  # % de perda estimada por ineficiência/dissipação
}

tudo_certo = False

while not tudo_certo:
    dados_usuario = {
        "energia": float(input("Energia atual (%): ")),
        "temperatura_externa": float(input("Temperatura externa: ")),
        "temperatura_interna": float(input("Temperatura interna: ")),
        "pressao_tanque": float(input("Pressao do tanque (bar): ")),
        "integridade_estrutural": int(input("Integridade Estrutural da nave:\nNegativo = 0 \nPositivo = 1: ")),
        "status_modulos_criticos": int(input("Status dos modulos criticos:\nFalha = 0 \nOK = 1: ")),
    }

    energia_kwh_disponivel = especificacoes_nave["capacidade_total_kwh"] * (dados_usuario["energia"] / 100)

    if (dados_usuario["energia"] >= dados_minimos["energia"]
        and dados_minimos["temperatura_externa"][0] <= dados_usuario["temperatura_externa"] <= dados_minimos["temperatura_externa"][1]
        and dados_minimos["temperatura_interna"][0] <= dados_usuario["temperatura_interna"] <= dados_minimos["temperatura_interna"][1]
        and dados_minimos["pressao_tanque"][0] <= dados_usuario["pressao_tanque"] <= dados_minimos["pressao_tanque"][1]
        and energia_kwh_disponivel >= especificacoes_nave["consumo_decolagem_kwh"]
        and dados_usuario["integridade_estrutural"] == dados_minimos["integridade_estrutural"]
        and dados_usuario["status_modulos_criticos"] == dados_minimos["status_modulos_criticos"]):
        print("\nTodos os requisitos foram atendidos.\nPRONTO PARA DECOLAR.")
        tudo_certo = True

    else:
        print("\nDECOLAGEM ABORTADA. Requisitos que falharam:")

        if dados_usuario["energia"] < dados_minimos["energia"]:
            print("- energia abaixo do minimo")
        if dados_usuario["temperatura_externa"] < dados_minimos["temperatura_externa"][0]:
            print("- temperatura externa abaixo do limite")
        if dados_usuario["temperatura_externa"] > dados_minimos["temperatura_externa"][1]:
            print("- temperatura externa acima do limite")
        if dados_usuario["temperatura_interna"] < dados_minimos["temperatura_interna"][0]:
            print("- temperatura interna abaixo do limite")
        if dados_usuario["temperatura_interna"] > dados_minimos["temperatura_interna"][1]:
            print("- temperatura interna acima do limite")
        if dados_usuario["pressao_tanque"] < dados_minimos["pressao_tanque"][0]:
            print("- pressao do tanque abaixo do limite")
        if dados_usuario["pressao_tanque"] > dados_minimos["pressao_tanque"][1]:
            print("- pressao do tanque acima do limite")
        if energia_kwh_disponivel < especificacoes_nave["consumo_decolagem_kwh"]:
            print("- energia insuficiente para decolagem")
        if dados_usuario["integridade_estrutural"] != dados_minimos["integridade_estrutural"]:
            print("- integridade estrutural comprometida")
        if dados_usuario["status_modulos_criticos"] != dados_minimos["status_modulos_criticos"]:
            print("- modulos criticos com falha")

        resposta = input("\nDeseja tentar novamente? (s/n): ").lower()
        if resposta != "s":
            print("Encerrando o programa.")
            break

if tudo_certo:
    print("\n=== Analise Energetica ===")

    carga_atual_percent = dados_usuario["energia"]

    energia_disponivel_kwh = especificacoes_nave["capacidade_total_kwh"] * (carga_atual_percent / 100)

    perda_kwh = energia_disponivel_kwh * (especificacoes_nave["perda_energetica_percent"] / 100)

    energia_liquida_kwh = energia_disponivel_kwh - perda_kwh

    energia_apos_decolagem_kwh = energia_liquida_kwh - especificacoes_nave["consumo_decolagem_kwh"]

    autonomia_horas = energia_apos_decolagem_kwh / especificacoes_nave["consumo_sistemas_kw"]

    print(f"\nEnergia disponivel (bruta): {energia_disponivel_kwh:.2f} kWh")
    print(f"Perdas energeticas ({especificacoes_nave['perda_energetica_percent']}%): {perda_kwh:.2f} kWh")
    print(f"Energia liquida: {energia_liquida_kwh:.2f} kWh")
    print(f"Energia apos decolagem: {energia_apos_decolagem_kwh:.2f} kWh")
    print(f"Autonomia estimada: {autonomia_horas:.2f} horas")
```
