```python
print("Bem-vindo ao sistema de decolagem!")

temperatura_externa = float(input("Temperatura externa atual: "))
temperatura_interna = float(input("Temperatura interna atual: "))
energia = float(input("Energia atual (%): "))

energia_segura = 50.0
energia_decolagem = 20.0
temperatura_segura = 25.0

if temperatura_externa >= temperatura_segura and temperatura_interna >= temperatura_segura:
    print("Temperatura segura")
    if energia >= energia_segura:
        print("Decolagem autorizada, você tem {:.2f}% kwh de energia".format(energia))
        print("Você vai ter um consumo de {:.2f}% kwh de energia para fazer a decolagem, vai ter ainda {:.2f}% kwh".format(
            energia_decolagem, energia - energia_decolagem))
    else:
        print("DECOLAGEM ABORTADA!!!! Energia insuficiente.")
else:
    print("DECOLAGEM ABORTADA!!!! Temperatura fora da faixa segura.")
```
